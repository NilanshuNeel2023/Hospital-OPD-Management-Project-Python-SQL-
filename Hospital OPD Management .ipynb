## Using Def Function : ##

```python
def generate_Paitent_id():
    query = """
    Select MAX(
            CAST(
                SUBSTRING(Paitent_id, 11) AS UNSIGNED
            )
        ) 
    from Paitent
    """
    cursor.execute(query)
    paitent = cursor.fetchone()
    
    if paitent[0] is None:
        New_number = 6475
    
    else:
        New_number = paitent[0] + 1
        
    Paitent_id = f"KDAH100104{New_number}"
    return Paitent_id

def New_paitent():
    
    Name = input("Enter the Name: ")
    Age = int(input("Age: "))
    DOB = input("Your DOB (YYYY-MM-DD): ")
    City = input("City/State: ")
    Contact = input("Contact Number:")
    Gender = input("Gender: ")
    Department = input("Department: ")
    Doctors = input("Doctors:")
    
    Paitent_id = generate_Paitent_id()
    
    print("Paitent ID:", Paitent_id)
    
    query = """
    INSERT INTO Paitent(Paitent_id, Name, Age, DOB, City, Contact, Gender, Department, Doctors)
    
    VALUES (%s,%s,%s,%s,%s,%s,%s,%s,%s)
    """
    
    values = (
        Paitent_id,
        Name,
        Age,
        DOB,
        City,
        Contact,
        Gender,
        Department,
        Doctors
    )
    
    cursor.execute(query, values)
    conn.commit()
    
    print("\nPaitent Registered Successfully!")
    print("Welcome:", Name)
    
def Follow_up():
    
    Paitent_id = input("Enter Paitent Registered-ID :")
    query = "Select * from Paitent where Paitent_id = %s"
    cursor.execute(query, (Paitent_id,))
    paitent = cursor.fetchone()
    
    if paitent:
        print("\nWelcome:", paitent[1])
        print("Age:", paitent[2])
        print("DOB:", paitent[3])
        print("City:", paitent[4])
        print("Contact:", paitent[5])
        print("Gender:",paitent[6])
        print("Department:",paitent[7])
        print("Doctor:",paitent[8])
    
    else:
        print("\n❌ID Not Found! Please Enter Valid ID")
        
def Bill_payment():
    
    Paitent_id = input("Enter Paitent Registred-ID: ")
    
    query = "Select * from paitent where paitent_id = %s"
    cursor.execute(query, (Paitent_id,))
    paitent = cursor.fetchone()
    
    if paitent:
        print("\nName:", paitent[1])
        print("Your Total Payable Bills: ")
        print("Thanks to Serve You... Keep Healthy😊!")
        print("\nThanks for your Visit🙂 !")

    else:
        print("\n❌ID Not Found! Please Enter Valid ID")
        

print("Welcome To City Hospital 🙂")
print("\nPRESS ON!")
print("\n1. New Patient")
print("2. Follow-Up Patient")
print("3. Bill Payment")

Choice = input("\nPlese Select the option(1/2/3): ")

if Choice == "1":
    New_paitent()

elif Choice == "2":
    Follow_up()
    
elif Choice == "3":
    Bill_payment()
    
    
else:
    print("Invalid Choice")
```
